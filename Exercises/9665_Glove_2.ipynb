### Load pretrained model

```python
%%time

# Download GloVe 300D model using Gensim
#  Trained on the Wikipedia 2014 + Gigaword 5 corpus
#  300-dimensional word vectors
#  Contains 400,000 vocabulary words
#  File size = 375 MB
glove_model = api.load("glove-wiki-gigaword-300")
```

### Use downloaded model

```python
# Access the vector for a word
glove_model['king']
```

```python
# Find words most similar to a given word
glove_model.most_similar('king', topn=5)
```

```python
# Solving analogies: man - woman + king = ?
analogy_result = glove_model.most_similar(positive=['king', 'woman'], negative=['man'], topn=5)
print(f"Analogy 'king - man + woman':", analogy_result)
```

```python
# Calculate similarity between 'king' and 'queen'
similarity = glove_model.similarity('king', 'queen')
print(f"Similarity between 'king' and 'queen':", similarity)
```

```python
# Create function to determine the odd one out
def find_odd_one_out(words):
    # Get word vectors for the words in the list
    vectors = [glove_model[word] for word in words if word in glove_model]
    
    if len(vectors) != len(words):
        missing_words = [word for word in words if word not in glove_model]
        return f"Missing words from the vocabulary: {missing_words}"

    # Calculate the average vector for all words
    avg_vector = np.mean(vectors, axis=0).reshape(1, -1)
    
    # Calculate similarity of each word to the average vector
    similarities = cosine_similarity(avg_vector, vectors).flatten()
    
    # Find the word with the least similarity to the average
    odd_word_idx = np.argmin(similarities)
    odd_word = words[odd_word_idx]
    
    return odd_word
```

```python
# Example: Animals and Fruits
words = ['dog', 'cat', 'apple', 'lion']
odd_word = find_odd_one_out(words)
print(f"The odd word out is: {odd_word}")
```

```python
# Example: Colors and Objects
words = ['red', 'blue', 'green', 'chair']
odd_word = find_odd_one_out(words)
print(f"The odd word out is: {odd_word}")
```

```python
# Example: Professions and Objects
words = ['doctor', 'teacher', 'pencil', 'engineer']
odd_word = find_odd_one_out(words)
print(f"The odd word out is: {odd_word}")
```
