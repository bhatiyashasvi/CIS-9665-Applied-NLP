- bert-base-nli-mean-tokens - is a sentence-transformer model, it maps sentences to a 768-dimensional vector space

## Part 2b

````python
# Create function to find similar sentences a similarity threshold
def find_similar_sentences(query_sentence, top_n=3, threshold=0.4):
    # Encode the query sentence
    query_sentence_embedding = model_2.encode(query_sentence, convert_to_tensor=True)

    # Compute cosine similarity with all sentence embeddings
    cosine_similarity_score = util.pytorch_cos_sim(query_sentence_embedding, sentence_embeddings_2)[0]

    # Get the top_n most similar indices
    top_n += 1  # To offset potential self-match
    top_results = np.argpartition(-cosine_similarity_score, range(top_n))[0:top_n]

    count = 0
    found_any = False
    for idx in top_results:
        score = cosine_similarity_score[idx].item()
        if score >= threshold:
            if not found_any:
                print(f'For query sentence [{query_sentence}], the most similar sentences above threshold {threshold} are:')
                found_any = True
            count += 1
            print(f'\t{count}. {sentences[idx]} : Similarity score = {score:.4f}')

    if not found_any:
        print(f'No sentences found with similarity above threshold {threshold:.2f}.')
````
