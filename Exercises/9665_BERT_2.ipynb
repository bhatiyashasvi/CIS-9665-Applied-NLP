````python
import pandas as pd
import numpy as np
from sentence_transformers import SentenceTransformer
from sentence_transformers import util
````

### Load data

````python
df = pd.read_csv('https://raw.githubusercontent.com/vjavaly/Baruch-CIS-9665/main/data/GameOfThrones.csv',
                 index_col='No. of Episode (Overall)')
df.shape
````

### Examine data

````python
df.head()
````

### Load sentence transformer model 'stsb-roberta-large'

````python
%%time

model = SentenceTransformer('stsb-roberta-large')
````

### Encode data (generate embeddings)

````python
# From dataframe column 'Title of the Episode', create list named 'titles_list'
titles_list = df['Title of the Episode'].tolist()

# From dataframe column 'Synopsis', create list named 'synopsis_list'
synopsis_list = df['Synopsis'].tolist()
````

````python
%%time

# Create embeddings for 'synopsis_list'
corpus_embedding = model.encode(synopsis_list, convert_to_tensor=True)
corpus_embedding
````

````python
# Create function to find similar episodes
def find_similar_episodes (query_title='Winter Is Coming', top_n=3):
    query_sentence=df[df['Title of the Episode']==query_title]['Synopsis'].values[0]
    query_sentence_embedding = model.encode(query_sentence, convert_to_tensor=True)
    cosine_similarity_score = util.pytorch_cos_sim(query_sentence_embedding, corpus_embedding)[0]
    top_n += 1
    top_results = np.argpartition(-cosine_similarity_score, range(top_n))[0:top_n]

    count=0
    for idx in top_results[0:top_n]:
        if count == 0:
            print(f'For query title [{query_title}], the most similar episodes are:')
        else:
            print(f'\t{count}. {titles_list[idx]} : Similarity score = {(cosine_similarity_score[idx]) * 100}')
        count += 1
````

````python
# Find similar episodes for title 'Winter Is Coming'
find_similar_episodes()
````

````python
# Find similar episodes for title 'Oathkeeper'
find_similar_episodes(query_title='Oathkeeper')
````

````python
# Find similar episodes for title 'Blood of My Blood'
find_similar_episodes(query_title='Blood of My Blood', top_n=5)
````
