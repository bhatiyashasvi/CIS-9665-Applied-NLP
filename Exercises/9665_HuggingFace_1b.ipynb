### Import libraries

```python
from google.colab import userdata
from huggingface_hub import login
from huggingface_hub import whoami
from huggingface_hub import hf_api
from pprint import pprint
import rich     # For rich text formatting
```

### Set up HuggingFace API token

```python
# Fetch token from Colab secrets
try:
    HF_TOKEN = userdata.get('HF_TOKEN')
    # Use token to log in to Hugging Face
    login(token=HF_TOKEN)
    print("Login successful!")
except KeyError:
    print("HF_TOKEN not found in Colab secrets. Please add it.")
except Exception as e:
    print(f"An error occurred during login: {e}")
```

```python
# Display info for user related to HF_TOKEN
pprint(whoami())
```

### List Hugging Face datasets and models

```python
# Connect to Hugging Face API
api = hf_api.HfApi()
```

```python
rich.inspect(api, methods=True)
```

```python
# List the first 10 models from the Hugging Face Hub
models = api.list_models(limit=10)

# Print the model names
for model in models:
    print(model.modelId)
```

```python
# Fetch model information
model_info = api.model_info('deepseek-ai/DeepSeek-V3-0324')

# Print the model info
pprint(model_info)
```

```python
# List the first 10 datasets from the Hugging Face Hub
datasets = list(api.list_datasets(limit=10))

# Print the dataset names
for dataset in datasets:
    print(dataset.id)
```

```python
# Fetch dataset information
dataset_info = api.dataset_info('Anthropic/EconomicIndex')

# Print the model info
pprint(dataset_info)
```
