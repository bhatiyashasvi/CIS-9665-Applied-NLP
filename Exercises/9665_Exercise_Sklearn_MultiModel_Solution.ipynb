### Import libraries

```python
import pandas as pd
import numpy as np
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.naive_bayes import GaussianNB
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score
```

### Load data

```python
# Read data from file (breast_cancer_diagnosis.csv) into dataframe
df = pd.read_csv('https://raw.githubusercontent.com/vjavaly/Baruch-CIS-9665/main/data/breast_cancer_diagnosis.csv')
```

### Examine data

```python
# Review dataframe shape
df.shape
```

```python
# Display first few rows of dataframe
df.head()
```

Note: NaN in radius column

### Prepare data

#### Drop "meaningless" variables
Remember to use "inplace=True"

```python
df.drop(["id", "name"], axis=1, inplace=True)
```

```python
# Display first few rows of updated dataframe
df.head()
```

#### Check for missing values

```python
df.isnull().sum()
```

#### Use the SimpleImputer to replace missing values

```python
imp_mean = SimpleImputer(missing_values=np.nan, strategy='mean')
```

```python
df = pd.DataFrame(imp_mean.fit_transform(df), columns=df.columns)
```

```python
# Display first few rows of updated dataframe
df.head()
```

#### Check for missing values again

```python
df.isnull().sum()
```

### Separate independent and dependent variables
* Independent variables: All remaining variables except Diagnosis
* Dependent variable: Diagnosis

```python
X = df.drop("diagnosis", axis=1)
y = df["diagnosis"]
```

#### Standardize the data

```python
scaler = StandardScaler()
X[X.columns] = scaler.fit_transform(X[X.columns])
X.head()
```

```python
# Display first few rows of dependent variable
y.head()
```

### Split data into training and test sets

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.3, random_state = 42)
```

### Train Logistic Regression model

```python
LogReg_model = LogisticRegression()
LogReg_model.fit(X_train, y_train)
```

### Evaluate Logistic Regression model

```python
LogReg_predictions = LogReg_model.predict(X_test)
LogReg_accuracy = accuracy_score(y_test, LogReg_predictions)
print("Logistic regression accuracy = ", round((LogReg_accuracy * 100), 5), "%")
```

### Train Random Forest Classifier model

```python
RF_model = RandomForestClassifier()
RF_model.fit(X_train, y_train)
```

### Evaluate Random Forest Classifier model

```python
RF_predictions = RF_model.predict(X_test)
RF_accuracy = accuracy_score(y_test, RF_predictions)
print("Random forest accuracy = ", round((RF_accuracy * 100), 5), "%")
```

### Train Naive Bayes model

```python
NBG_model = GaussianNB()
NBG_model.fit(X_train, y_train)
```

### Evaluate Naive Bayes model

```python
NBG_predictions = NBG_model.predict(X_test)
NBG_accuracy = accuracy_score(y_test, NBG_predictions)
print("Naive Bayes (Gaussian) accuracy = ", round((NBG_accuracy * 100), 5), "%")
```

### Train Support Vector Classifier model

```python
SVC_model = SVC()
SVC_model.fit(X_train, y_train)
```

### Evaluate Support Vector Classifier model

```python
SVC_predictions = SVC_model.predict(X_test)
SVC_accuracy = accuracy_score(y_test, SVC_predictions)
print("Support vector machine accuracy = ", round((SVC_accuracy * 100), 5), "%")
```
