### Write clear and specific instructions

````python
# Instantiate OpenAI client
client = OpenAI()

# Specify OpenAI model
OpenAI_model = "gpt-4o-mini"
````

````python
# Set up chat
completion = client.chat.completions.create(
  model=OpenAI_model,
  messages=[
    {"role": "user",
     "content": "Who won the FIFA Mens Soccer World Cup in 2022?"}
  ]
)
````

````python
print(completion)
````

````python
pprint(completion.choices[0].message.content)
````

````python
completion = client.chat.completions.create(
  model=OpenAI_model,
  messages=[
    {"role": "system",
     "content": "You are a helpful assistant."},
    {"role": "user",
     "content": "Who won the FIFA Mens Soccer World Cup in 2022?"}
  ]
)
````

````python
pprint(completion.choices[0].message.content)
````

````python
# Create chat function
def chat_function_1(prompt):
    client = OpenAI()
    OpenAI_model = "gpt-4o-mini"

    completion = client.chat.completions.create(
        model=OpenAI_model,
        messages=[
            {"role": "system",
             "content": "You are an expert in sports history."},
            {"role": "user",
             "content": prompt}
        ]
    )
    print(completion.choices[0].message.content)
````

````python
chat_function_1("Who won the most recent FIFA Womens Soccer World Cup")
````

````python
chat_function_1("Who are the last 5 winners of FIFA Mens Soccer World Cup?  List the country and the year.")
````

````python
prompt = """
"Who are the last 5 winners of FIFA Mens Soccer World Cup?
List the winning country, the losing country, the year and the location.
Return the output in json format."
"""

chat_function_1(prompt)
````

````python
prompt = """
"Who are the last 5 winners of FIFA Womens Soccer World Cup?
List the winning country, the losing country, the year and the location.
Return the output in tabular format."
"""

chat_function_1(prompt)
````

````python
prompt = """
"Gather the movie Oscars results between 2009 and 2013 for best picture, best male actor, best female actor and year?
Return the output in a prose format.
The output should be in paragraphs with a title for each paragraph.
Each sentence in the paragraphs should be displayed in lines of 120 characters"
"""

chat_function_1(prompt)
````

````python
text = """
You should express what you want a model to do by providing instructions that are as clear
and specific as you can possibly make them. This will guide the model towards the desired output, and
reduce the chances of receiving irrelevant or incorrect responses. Don't confuse writing a clear
prompt with writing a short prompt. In many cases, longer prompts provide more clarity and context
for the model, which can lead to more detailed and relevant outputs.
"""

prompt = f"""
Summarize the text delimited by triple backticks into a single sentence of less than 120 characters.
```{text}```
"""

chat_function_1(prompt)
````

## Part 2

### Review impacts of different parameter combinations

#### Example 1: Expect to have the similar discrete answers

````python
# Create new chat function
#  Pass first combination of parameters
def chat_function_2(PROMPT):
    client = OpenAI()

    response = client.chat.completions.create(
        model='gpt-4o-mini',
        temperature=0,
        max_tokens=100,
        top_p=0.1,
        frequency_penalty=0,
        presence_penalty=0,
        messages=[
            {"role": "system",
             "content": "You are a helpful assistant."},
            {"role": "user",
             "content": PROMPT}
        ]
    )
    print(response.choices[0].message.content)
````

````python
prompt_2 = """Describe a futuristic city in the year 2150.
The output should be prose.
Each sentence should be displayed in lines of 80 characters"
"""
````

````python
# Example 1 model output
chat_function_2(prompt_2)
````

````python
# Example 1 model output
chat_function_2(prompt_2)
````

#### Example 2: Expect to have different, more random, answers with more sophisticated words

````python
# Create new chat function
#  Pass second combination of parameters
def chat_function_3(PROMPT):
    client = OpenAI()

    response = client.chat.completions.create(
        model='gpt-4o-mini',
        temperature=0.7,
        max_tokens=100,
        top_p=0.1,
        frequency_penalty=2,
        presence_penalty=0,
        messages=[
            {"role": "system",
             "content": "You are a helpful assistant."},
            {"role": "user",
             "content": PROMPT}
        ]
    )
    print(response.choices[0].message.content)
````

````python
# Example 2 model output
chat_function_3(prompt_2)
````

````python
# Example 2 model output
chat_function_3(prompt_2)
````

#### Example 3: Expect to have a completely new answers
