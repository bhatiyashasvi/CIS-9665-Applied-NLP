### Sentiment analysis

```python
%%time

# Initialize the sentiment analysis pipeline
sentiment_analyzer = pipeline("sentiment-analysis",
                              model="distilbert/distilbert-base-uncased-finetuned-sst-2-english",
                              revision="714eb0f")
```

```python
pd.set_option('display.max_colwidth', None)
```

```python
# Simple sentences
sa_simple_sentences = [
    "I love this movie! It's so amazing.",
    "The food was terrible, and the service was awful.",
    "This is a neutral statement.",
    "I'm feeling quite happy today.",
    "The news made me incredibly sad.",
    "What a wonderful day!",
    "I have mixed feelings about this.",
    "The product was okay, but not great.",
    "This is the best thing that's ever happened!",
    "I'm so disappointed.",
]
```

```python
# Perform sentiment analysis
results = sentiment_analyzer(sa_simple_sentences)

# Store results in dataframe
df = pd.DataFrame({
    "sentence": sa_simple_sentences,
    "sentiment": [result["label"] for result in results],
    "score": [result["score"] for result in results],
})

print("Simple sentences results")
df
```

```python
# More Complex Examples
sa_complex_sentences = [
    "Despite the initial setbacks, the team managed to pull through and deliver an outstanding performance.",
    "While the concept is intriguing, the execution leaves much to be desired.",
    "The sheer brilliance of the artwork was only slightly marred by the poor lighting.",
    "Although the movie had its moments, the overall plot felt rather convoluted.",
    "Even though the weather was dreadful, the atmosphere at the concert was electric.",
]
```

```python
# Perform sentiment analysis
complex_results = sentiment_analyzer(sa_complex_sentences)

# Store results in dataframe
complex_df = pd.DataFrame({
    "sentence": sa_complex_sentences,
    "sentiment": [result["label"] for result in complex_results],
    "score": [result["score"] for result in complex_results],
})

print("Complex sentences results")
complex_df
```

### Named entity recognition (NER)

```python
%%time

# Initialize the NER pipeline
ner_recognizer = pipeline("ner",
                          model='dslim/bert-base-NER')
```

```python
# Example simple sentences for NER
ner_simple_sentences = [
    "Apple Inc. is headquartered in Cupertino, California.",
    "Barack Obama was the 44th president of the United States.",
    "I'm going to visit the Eiffel Tower in Paris next week.",
    "Google was founded by Larry Page and Sergey Brin.",
    "The Mona Lisa is displayed at the Louvre Museum in France.",
    "Angela Merkel served as the Chancellor of Germany.",
    "Amazon's headquarters are located in Seattle, Washington.",
    "The movie 'The Godfather' was directed by Francis Ford Coppola.",
    "Mount Everest is the highest mountain in the world.",
    "Leonardo da Vinci was a famous Italian artist.",
]
```

```python
# Perform NER
results = ner_recognizer(ner_simple_sentences)
```

```python
# Store results in dataframe
data = []
for sentence, result in zip(ner_simple_sentences, results):
    entities = []
    for entity in result:
        entities.append({
            "entity": entity["entity"],
            "word": entity["word"],
            "start": entity["start"],
            "end": entity["end"],
            "score": entity["score"]
            })
    data.append({"sentence": sentence, "entities": entities})

df = pd.DataFrame(data)

print("NER simple results")
df
```

```python
# More complex sentences for NER
ner_complex_sentences = [
    "On July 20, 1969, Neil Armstrong and Buzz Aldrin became the first humans to land on the moon.",
    "The company announced a new partnership with IBM at the Consumer Electronics Show in Las Vegas.",
    "During World War II, Winston Churchill served as the Prime Minister of the United Kingdom.",
    "The painting 'Starry Night' by Vincent van Gogh is a masterpiece of post-impressionism.",
    "The Olympic Games were held in Tokyo, Japan, in 2021, after a year's delay due to the COVID-19 pandemic.",
]
```

```python
# Perform NER
results = ner_recognizer(ner_complex_sentences)
```

```python
# Store results in dataframe
data = []
for sentence, result in zip(ner_complex_sentences, results):
    entities = []
    for entity in result:
        entities.append({
            "entity": entity["entity"],
            "word": entity["word"],
            "start": entity["start"],
            "end": entity["end"],
            "score": entity["score"]
            })
    data.append({"sentence": sentence, "entities": entities})

df = pd.DataFrame(data)

print("NER complex results")
df
```

### Question Answering

```python
%%time

# Initialize the question answering pipeline with a specific model
question_answerer = pipeline("question-answering",
                             model="distilbert/distilbert-base-cased-distilled-squad")
```

```python
# Context and questions (simple)
context = """
The Golden State Warriors are an American professional basketball team based in San Francisco.
The Warriors compete in the National Basketball Association (NBA), as a member of the league's
Western Conference Pacific Division. Founded in 1946 in Philadelphia, the Warriors moved to
the San Francisco Bay Area in 1962 and then to San Francisco in 2019. The team plays its home
games at the Chase Center. The Warriors won the 2022 NBA Finals, their seventh NBA championship.
"""

questions = [
    "Where are the Golden State Warriors based?",
    "Which league do the Warriors compete in?",
    "When were the Warriors founded?",
    "Where do the Warriors play their home games?",
    "How many NBA championships have the Warriors won?",
    "What year did the warriors win their last championship?",
]
```

```python
# Perform question answering
results = []
for question in questions:
    result = question_answerer(question=question, context=context)
    results.append(result)
```

```python
# Create a Pandas DataFrame for better visualization
df = pd.DataFrame({
    "question": questions,
    "answer": [result["answer"] for result in results],
    "start": [result["start"] for result in results],
    "end": [result["end"] for result in results],
    "score": [result["score"] for result in results],
})

df
```

```python
# Context and questions (complex)
complex_context = """
The Amazon rainforest, covering much of northwestern Brazil and extending into Colombia, Peru and other South American countries,
is the world’s largest tropical rainforest, famed for its biodiversity. It’s crisscrossed by thousands of rivers, including the powerful
Amazon. The region contains over 3 million species of plants and animals, and 1 million indigenous people.
"""

complex_questions = [
    "Where is the Amazon rainforest primarily located?",
    "What is the Amazon rainforest known for?",
    "How many species of plants and animals are estimated to live in the Amazon?",
    "Who are the original inhabitants of the Amazon rainforest?",
    "What is the name of the largest river that crosses the Amazon?",
]
```

```python
# Perform question answering
complex_results = []
for question in complex_questions:
    result = question_answerer(question=question, context=complex_context)
    complex_results.append(result)
```

```python
# Create a Pandas DataFrame for better visualization
complex_df = pd.DataFrame({
    "question": complex_questions,
    "answer": [result["answer"] for result in complex_results],
    "start": [result["start"] for result in complex_results],
    "end": [result["end"] for result in complex_results],
    "score": [result["score"] for result in complex_results],
})

complex_df
```
